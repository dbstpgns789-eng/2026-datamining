파일을 sc.textFile()로 읽어서 다음을 계산하세요.

• 첫 번째 정수 쌍의 첫 번째 원소

• 모든 정수 쌍의 첫 번째 원소 중 최댓값

• 모든 정수 쌍의 첫 번째 원소 중 최솟값

• 첫 번째 원소가 1인 정수 쌍의 개수

• 첫 번째 원소가 2인 정수 쌍의 개수

이 다섯 개의 숫자를 모두 더한 뒤 출력하세요.

```python
from pyspark import SparkContext
import urllib.request

sc = SparkContext.getOrCreate()
url = "https://raw.githubusercontent.com/nongaussian/class-2024-datamining/main/quiz/int-pair-input.txt"
urllib.request.urlretrieve(url, "int-pair-input.txt")
```

```python
data = sc.textFile("int-pair-input.txt")
data.take(10)
```

```python
pairs = data.map(lambda line: tuple(map(int, line.strip("()").split(", "))))
pairs.take(10)
```

```python
first_value = pairs.first()[0]  #첫 번째 정수 쌍의 첫 번째 원소
print(first_value)
```

```python
first_elements = pairs.map(lambda x: x[0])

max_value = first_elements.max()  #모든 정수 쌍의 첫 번째 원소 중 최댓값
min_value = first_elements.min()  #모든 정수 쌍의 첫 번째 원소 중 최솟값

print(max_value)
print(min_value)
```

```python
fe_1 = pairs.filter(lambda x: x[0] == 1).count()
fe_2 = pairs.filter(lambda x: x[0] == 2).count()

print(fe_1) # 첫 번째 원소가 1인 정수 쌍의 개수
print(fe_2) # 첫 번째 원소가 2인 정수 쌍의 개수
```

```python
result = first_value + max_value + min_value + fe_1 + fe_2
print(result)
```
