```python
from pyspark import SparkContext
import urllib.request
```

```python
url = "https://raw.githubusercontent.com/nongaussian/class-2024-datamining/main/quiz/sherlock.txt"

urllib.request.urlretrieve(url, "sherlock.txt")
```

```python
sc = SparkContext.getOrCreate()

data = sc.textFile("sherlock.txt")

data.take(5)
```

```python
words = data.flatMap(lambda line: line.split())
words.take(5)
```

```python
valid_words = words.filter(lambda word: len(word) > 2)
valid_words.take(5)
```

```python
head_words = valid_words.map(lambda word: word.lower()[:2])
head_words.take(5)
```

```python
pairs = head_words.map(lambda word: (word, 1))
pairs.take(5)
```

```python
counts = pairs.reduceByKey(lambda a, b: a + b)
counts.take(5)
```

```python
small_counts = counts.filter(lambda x: x[1] < 32)
small_counts.take(5)
```

```python
result = small_counts.map(lambda x: x[1]).sum()
print(result)
```
