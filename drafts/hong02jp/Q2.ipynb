Step 2 : Spark 초기화

```python
import wget
from pyspark.sql import SparkSession

# SparkSession 및 SparkContext 초기화
spark = SparkSession.builder.master("local[*]").appName("HW1_Q2").getOrCreate()
sc = spark.sparkContext
```

Step 3 : 데이터 로드 및 전처리

```python
# 1. 정수 쌍 텍스트 파일 다운로드
url_q2 = "https://raw.githubusercontent.com/nongaussian/class-2024-datamining/main/quiz/int-pair-input.txt"
wget.download(url_q2, "int-pair-input.txt")

# 2. RDD 생성 및 첫 번째 원소만 정수형으로 추출
rdd_q2 = sc.textFile("int-pair-input.txt")
first_elements = rdd_q2.map(
    lambda line: int(line.replace('(', '').replace(')', '').replace(',', ' ').split()[0])
)

# 이후 5개의 Action이 동일한 RDD를 참조하므로, 중복 연산을 방지하고 속도를 높이기 위해 데이터를 메모리에 캐싱
first_elements.cache()
```

Step 4 : 5가지 연산 수행

```python
# 1. 첫 번째 정수 쌍의 첫 번째 원소
val1 = first_elements.first()

# 2. 모든 정수 쌍의 첫 번째 원소 중 최댓값 (reduce 활용)
val2 = first_elements.reduce(lambda a, b: a if a > b else b)

# 3. 모든 정수 쌍의 첫 번째 원소 중 최솟값 (reduce 활용)
val3 = first_elements.reduce(lambda a, b: a if a < b else b)

# 4. 첫 번째 원소가 1인 정수 쌍의 개수
val4 = first_elements.filter(lambda x: x == 1).count()

# 5. 첫 번째 원소가 2인 정수 쌍의 개수
val5 = first_elements.filter(lambda x: x == 2).count()
```

Step 5 : 최종 결과 병합 및 출력

```python
# 결과 병합
final_sum = val1 + val2 + val3 + val4 + val5

# 결과 검증 및 최종 출력
print("1. 첫 번째 원소:", val1)
print("2. 최댓값:", val2)
print("3. 최솟값:", val3)
print("4. 값이 1인 개수:", val4)
print("5. 값이 2인 개수:", val5)
print("-" * 35)
print("다섯 개의 숫자를 모두 더한 결과:", final_sum)
```
