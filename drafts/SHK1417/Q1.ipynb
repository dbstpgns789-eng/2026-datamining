```python
from pyspark.sql import SparkSession

spark = SparkSession.builder\
        .master("local")\
        .appName("Q1")\
        .getOrCreate()
sc = spark.sparkContext
```

```python
import wget

wget.download(
    "https://raw.githubusercontent.com/nongaussian/class-2024-datamining/main/quiz/sherlock.txt",
    "sherlock.txt"
)

rdd = sc.textFile("sherlock.txt")
```

```python
result = (rdd.flatMap(lambda txt: txt.lower().split(' '))  #각 단어 분리 & 대소문자 구분X
      .filter(lambda x: len(x) > 2)                        # 길이가 2보다 큰 단어만 고려
      .map(lambda x: (x[:2], 1))                           # 각 단어의 앞 두글자(2-head word)
      .reduceByKey(lambda x, y: x+y)                       # 2-head word들의 빈도 합
      .filter(lambda x: x[1] < 32)                         # 빈도가 32 미만
      .map(lambda x: x[1])                                 # 빈도값 추출
      .sum())                                              # 합계 계산
```

```python
print(result)
```
