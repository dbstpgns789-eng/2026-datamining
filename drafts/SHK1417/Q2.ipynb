```python
import wget

wget.download(
    "https://raw.githubusercontent.com/nongaussian/class-2024-datamining/main/quiz/int-pair-input.txt",
    "int-pair-input.txt"
)

data = sc.textFile("int-pair-input.txt")
```

```python
num = data.map(lambda x: eval(x))  # 문자열 튜플로 바꿈

# 첫 번째 정수 쌍의 첫 번째 원소
num1 = num.first()[0]
# 모든 정수 쌍의 첫 번째 원소 중 최댓값
num2 = num.map(lambda x: x[0]).reduce(lambda a, b: a if a > b else b)
# 모든 정수 쌍의 첫 번째 원소 중 최솟값
num3 = num.map(lambda x: x[0]).reduce(lambda a, b: a if a < b else b)
# 첫 번째 원소가 1인 정수 쌍의 개수
num4 = num.filter(lambda x: x[0] == 1).count()
# 첫 번째 원소가 2인 정수 쌍의 개수
num5 = num.filter(lambda x: x[0] == 2).count()
```

```python
total = num1 + num2 + num3 + num4 + num5  # 다섯 개의 숫자의 총합
print(total)
```
